```python
import pandas as pd
import numpy as np

# 创建一个包含正数和NaN的Series
s = pd.Series([1, 2, np.nan, 4, 5, np.nan, 7])

# 计算50%分位数
median = s.quantile(0.5)
print("Median:", median)

# 计算25%分位数
first_quartile = s.quantile(0.25)
print("First Quartile:", first_quartile)

# 计算75%分位数
third_quartile = s.quantile(0.75)
print("Third Quartile:", third_quartile)
```

```python
import numpy as np
import pandas as pd
import timeit  # 导入timeit模块

# 创建一个大的DataFrame进行测试
df = pd.DataFrame({
    'x': np.random.randn(1000000),
    'y': np.random.randn(1000000),
    'z': np.random.randn(1000000)
})

# 定义要测试的两种写法
def test_method_1():
    return np.where((df.x > df.y) | (df.x < df.z), 0, 1)

def test_method_2():
    return np.where(df.x > 0, np.where(df.x > df.y, 0, 1), np.where(df.x < df.z, 0, 1))

# 测试第一种写法
print(timeit.timeit('test_method_1()', globals=globals(), number=10))

# 测试第二种写法
print(timeit.timeit('test_method_2()', globals=globals(), number=10))
```
