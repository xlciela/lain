```python
import pandas as pd

# 创建一个简单的DataFrame
df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6],
    'C': [7, 8, 9]
})

# 应用函数于每一列，计算列的总和
column_sums = df.apply(lambda column: column.sum())
print(column_sums)
```
