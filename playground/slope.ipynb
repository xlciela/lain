```python
import pandas as pd
```

```python
import numpy as np
import talib.abstract as ta
from scipy import stats
```

```python
tdf = pd.read_csv('bitcoin.csv').tail(100)
```

```python
tdf.columns = [col.lower() for col in tdf.columns]
tdf
```

```python
tdf['ma20'] = ta.EMA(tdf['close'], 20) 
```

```python
tdf.tail(10)
```

```python
def calc_slope(x):
    return np.polyfit(range(5), x, 1)[0]
```

```python
tdf['slope'] = tdf['ma20'].rolling(5).apply(calc_slope, raw= True)
```

```python
tdf
```

```python
tdf['slope2'] = tdf['ma20'].rolling(5).apply(lambda x: stats.linregress(range(5), x.values )[0])
```

```python
tdf
```

```python
tdf['slope3'] = ta.LINEARREG_SLOPE(tdf['ma20'], timeperiod=5)
```
