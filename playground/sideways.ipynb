```python
import pandas as pd
```

```python
import numpy as np
import talib.abstract as ta
```

```python
tdf = pd.read_csv('bitcoin.csv').tail(100)
```

```python
tdf.columns = [col.lower() for col in tdf.columns]
```

```python
tdf
```

```python
tdf['top'] = ta.EMA(tdf.high, 20)
tdf['bottom'] = ta.EMA(tdf.low, 20)
```

```python
tdf
```

```python
tdf['within_range'] = np.where(
    (tdf['close'] < tdf['top']) & (tdf['close'] > tdf['bottom']), True, False
)
```
