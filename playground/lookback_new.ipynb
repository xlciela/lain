```python
import pandas as pd
from pandas import Series
from typing import Callable
```

```python
s1 = pd.Series([1, 2, 3, 4, 5,2,3,1,8,3,8,138])
s2 = pd.Series([1, 4, 3, 5,1,22,13,45,42,144,89, 8888])
```

```python
len(s1)
```

```python
s1-s2
```

```python
def lookback_cond(s1:Series, s2: Series, callback:Callable[[Series, Series], Series], timeperiod: int)-> Series:
    s = s1-s2
    result =  s.rolling(window=timeperiod).apply(callback, raw= False).fillna(0).astype(bool)
    print(result)
    return result
```

```python
cond = lookback_cond(s1, s2, lambda x: x.max()>0 , 2)
```

```python
cond.rolling(window=3).apply(lambda x: x.any()).fillna(0)
```
