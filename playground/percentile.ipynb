```python
import numpy as np
def custom_func(x):
    # 忽略NaN值
    print('hello')
    not_nan = ~np.isnan(x)
    if np.sum(not_nan) == 0:
        return np.nan  # 如果窗口中全是NaN，则返回NaN
    # 使用NumPy的索引方式获取最后一个元素
    last_value = x[-1]
    print(last_value)
    return np.sum(x[not_nan] < last_value) / np.sum(not_nan)
```

```python
# 假设df是您的DataFrame，timeperiod是滚动窗口的大小
# 首先使用mask函数将不满足条件的值替换为NaN

# 定义一个计算标准差的函数，忽略NaN值
def rolling_nanstd(x):
    return np.nanstd(x, ddof=0)

# 使用rolling().apply()结合自定义的rolling_nanstd函数来计算标准差，忽略NaN
std_series = test.rolling(window=5, min_periods=1).apply(rolling_nanstd, raw=False)

# 结果存储在std_series中
```
