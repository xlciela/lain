```python
from abc import ABCMeta, abstractmethod 
import pandas as pd
import numpy as np

class Strategy(metaclass=ABCMeta): 
    def __init__(self, data):
        self.data = data

    def define_strategy():
        df = self.data.loc[self.data.Complete == True].copy()
        # ************** Define Strategy **************
        df = df[['close', 'volume']].copy()
        df['returns'] = np.log(df.close / df.close.shift(1))
        df['vol_ch'] = np.log(df.volume.div(df.volume.shift(1)))
        df.loc[df.vol_ch > 3, 'vol_ch'] = np.nan
        df.loc[df.vol_ch < -3, 'vol_ch'] = np.nan

        cond1 = df.returns < self.returns_threshold[0]
        cond2 = df.vol_ch.between(self.vol_threshold[0], self.vol_threshold[1])
        cond3 = df.returns > self.returns_threshold[1]

        df['position'] = 0
        df.loc[cond1 & cond2, 'position'] = 1
        df.loc[cond3 & cond2, 'position'] = -1
        #**********************************************
        
        self.prepared_data = df.copy()

    def execute_trades(self): 
        if self.prepared_data.position.iloc[-1] == 1: # long signal: go/stay long
            if self.position == 0:
                self.exchange.create_market_buy_order(self.symbol, self.amount)
            elif self.position ==-1:
                self.exchange.create_market_buy_order(self.symbol, self.amount*2)
            self.position = 1
        elif self.prepared_data.position.iloc[-1] == -1: # short signal: go/stay short
            pass
        if self.prepared_data.position.iloc[-1] == 0: # neutral signal: go flat
            pass
            
                
            


  
    
```
